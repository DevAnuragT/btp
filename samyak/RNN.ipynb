```python
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.preprocessing import StandardScaler

from tensorflow.keras import Model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    SimpleRNN,
    Lambda,
    Concatenate,
    Dense,
    Dropout,
)
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.metrics import AUC
from tensorflow.keras.optimizers import Adam, RMSprop

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
)
```

```python
if 'combined_df' not in locals():
    combined_df = pd.read_parquet("combined_dataset.parquet")

combined_df = combined_df.copy()

combined_df["dt"] = pd.to_datetime(combined_df["dt"])
```

```python
SERIES_COLUMNS = ["city_id", "store_id", "product_id"]
FEATURE_COLUMNS = [
    "sale_amount",
    "stock_hour6_22_cnt",
    "discount",
    "holiday_flag",
    "activity_flag",
    "precpt",
    "avg_temperature",
    "avg_humidity",
    "avg_wind_level",
    "hours_sale_sum",
    "hours_sale_max",
    "hours_stock_status_sum",
    "hours_stock_status_mean",
]
TARGET_COLUMNS = [f"target_stock_h{hour:02d}" for hour in range(24)]


def to_24_float_list(value, length=24):
    if isinstance(value, np.ndarray):
        values = value.tolist()
    elif isinstance(value, (list, tuple)):
        values = list(value)
    else:
        values = []

    values = [0.0 if pd.isna(item) else float(item) for item in values[:length]]
    if len(values) < length:
        values.extend([0.0] * (length - len(values)))
    return values


def add_series_id(data):
    data = data.copy()
    data["series_id"] = (
        data["city_id"].astype(str)
        + "_"
        + data["store_id"].astype(str)
        + "_"
        + data["product_id"].astype(str)
    )
    return data


def clean_frame_like_reference(data):
    data = add_series_id(data)
    data["dt"] = pd.to_datetime(data["dt"], errors="coerce")
    data = data.dropna(subset=["dt", *SERIES_COLUMNS])
    data = data.drop_duplicates(subset=["series_id", "dt"], keep="last")

    stock_matrix = np.array(
        data["hours_stock_status"].apply(to_24_float_list).tolist(),
        dtype=np.float32,
    )
    sales_matrix = np.array(
        data["hours_sale"].apply(to_24_float_list).tolist(),
        dtype=np.float32,
    )

    for hour in range(24):
        data[f"stock_h{hour:02d}"] = stock_matrix[:, hour]

    data["hours_sale_sum"] = sales_matrix.sum(axis=1)
    data["hours_sale_max"] = sales_matrix.max(axis=1)
    data["hours_stock_status_sum"] = stock_matrix.sum(axis=1)
    data["hours_stock_status_mean"] = data["hours_stock_status_sum"] / 24.0

    for column in FEATURE_COLUMNS:
        data[column] = pd.to_numeric(data[column], errors="coerce").fillna(0.0)

    return data.sort_values(["series_id", "dt"]).reset_index(drop=True)


def add_next_day_targets_like_reference(data):
    data = data.copy()
    stock_columns = [f"stock_h{hour:02d}" for hour in range(24)]
    for source_col, target_col in zip(stock_columns, TARGET_COLUMNS):
        data[target_col] = data.groupby("series_id")[source_col].shift(-1)
    return data.dropna(subset=TARGET_COLUMNS).reset_index(drop=True)

combined_df = clean_frame_like_reference(combined_df)
combined_df = add_next_day_targets_like_reference(combined_df)
combined_df[["series_id", "dt", "hours_stock_status_sum", *TARGET_COLUMNS[:4]]].head()
```

```python
combined_df["avg_hourly_sales"] = combined_df["hours_sale"].apply(np.mean)

combined_df["max_hourly_sales"] = combined_df["hours_sale"].apply(np.max)

combined_df["std_hourly_sales"] = combined_df["hours_sale"].apply(np.std)
```

```python
feature_cols = FEATURE_COLUMNS
target_cols = TARGET_COLUMNS
TARGET_HOURS = 24

DEMAND_FEATURES = [
    "sale_amount",
    "discount",
    "holiday_flag",
    "hours_sale_sum",
]
INVENTORY_FEATURES = [
    "stock_hour6_22_cnt",
    "hours_stock_status_sum",
]

demand_indices = [feature_cols.index(feature) for feature in DEMAND_FEATURES]
inventory_indices = [feature_cols.index(feature) for feature in INVENTORY_FEATURES]
```

```python
scaler = StandardScaler()
train_mask = combined_df["data_split"].eq("train")

scaler.fit(combined_df.loc[train_mask, feature_cols])
combined_df[feature_cols] = scaler.transform(combined_df[feature_cols])
```

```python
LOOKBACK = 14
TRAIN_DAYS = 75
TARGET_HOURS = 24
```

```python
X_train = []
y_train = []

X_test = []
y_test = []
sequence_meta_train = []
sequence_meta_test = []

for series_id, group in combined_df.groupby("series_id", sort=False):
    group = group.sort_values("dt").reset_index(drop=True)

    if len(group) <= LOOKBACK:
        continue

    features = group[feature_cols].to_numpy(dtype=np.float32)
    targets = group[target_cols].to_numpy(dtype=np.float32)
    dates = group["dt"].to_numpy()

    # Match the reference file: features[end-LOOKBACK:end] predict target day dates[end].
    # Because target columns are shifted, targets[end-1] is the next day's 24-hour vector.
    for end in range(LOOKBACK, len(group)):
        sample_x = features[end - LOOKBACK:end]
        sample_y = targets[end - 1]
        target_split = group.loc[end, "data_split"]
        meta = {"series_id": series_id, "target_date": str(pd.Timestamp(dates[end]).date())}

        if target_split == "train":
            X_train.append(sample_x)
            y_train.append(sample_y)
            sequence_meta_train.append(meta)
        elif target_split == "test":
            X_test.append(sample_x)
            y_test.append(sample_y)
            sequence_meta_test.append(meta)
```

```python
X_train = np.array(X_train, dtype=np.float32)
y_train = np.array(y_train, dtype=np.float32)

X_test = np.array(X_test, dtype=np.float32)
y_test = np.array(y_test, dtype=np.float32)
```

```python
print("Training Variation 1: 24-Hour SimpleRNN Baseline")
model = Sequential([
    Input(shape=(LOOKBACK, X_train.shape[2])),
    SimpleRNN(32, activation="tanh"),
    Dense(TARGET_HOURS, activation="sigmoid"),
])
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["binary_accuracy", AUC(name="auc"), AUC(name="pr_auc", curve="PR")],
)
history = model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=20,
    batch_size=64,
    verbose=1,
)
```

```python
def print_24h_metrics(model, X_eval, y_eval, threshold=0.5, name="Model"):
    loss, binary_accuracy, auc, pr_auc = model.evaluate(X_eval, y_eval, verbose=1)
    probabilities = model.predict(X_eval, batch_size=256, verbose=1)
    predictions = (probabilities >= threshold).astype(int)

    labels_flat = y_eval.astype(int).ravel()
    predictions_flat = predictions.ravel()

    print(f"{name} Test Loss:", loss)
    print(f"{name} Hour-Level Binary Accuracy @ {threshold}:", binary_accuracy)
    print(f"{name} Hour-Level Precision @ {threshold}:", precision_score(labels_flat, predictions_flat, zero_division=0))
    print(f"{name} Hour-Level Recall @ {threshold}:", recall_score(labels_flat, predictions_flat, zero_division=0))
    print(f"{name} Hour-Level F1 @ {threshold}:", f1_score(labels_flat, predictions_flat, zero_division=0))
    print(f"{name} ROC AUC:", auc)
    print(f"{name} PR AUC:", pr_auc)
    print(f"{name} Exact 24H Match Rate:", np.mean(np.all(predictions == y_eval.astype(int), axis=1)))
    print(f"{name} Mean Absolute Hour Count Error:", np.mean(
        np.abs(predictions.sum(axis=1) - y_eval.astype(int).sum(axis=1))
    ))

print_24h_metrics(model, X_test, y_test, threshold=0.5, name="SimpleRNN Baseline")
```

```python
def evaluate_24h_predictions(model, X_eval, y_eval, threshold=0.5, batch_size=256):
    probabilities = model.predict(X_eval, batch_size=batch_size, verbose=1)
    predictions = (probabilities >= threshold).astype(int)
    labels_flat = y_eval.astype(int).ravel()
    predictions_flat = predictions.ravel()

    return {
        "threshold": threshold,
        "hour_level_accuracy": accuracy_score(labels_flat, predictions_flat),
        "hour_level_precision": precision_score(labels_flat, predictions_flat, zero_division=0),
        "hour_level_recall": recall_score(labels_flat, predictions_flat, zero_division=0),
        "hour_level_f1": f1_score(labels_flat, predictions_flat, zero_division=0),
        "exact_24h_match_rate": np.mean(np.all(predictions == y_eval.astype(int), axis=1)),
        "mean_absolute_hour_count_error": np.mean(
            np.abs(predictions.sum(axis=1) - y_eval.astype(int).sum(axis=1))
        ),
    }

evaluate_24h_predictions(model, X_test, y_test, threshold=0.5)
```

```python
def build_dual_stream_rnn_24h_model(
    lookback,
    n_features,
    demand_indices,
    inventory_indices,
    target_hours=24,
):
    inputs = Input(shape=(lookback, n_features))

    demand_stream = Lambda(lambda x: tf.gather(x, demand_indices, axis=-1))(inputs)
    inventory_stream = Lambda(lambda x: tf.gather(x, inventory_indices, axis=-1))(inputs)

    demand_branch = SimpleRNN(16, activation="tanh", dropout=0.10)(demand_stream)
    inventory_branch = SimpleRNN(16, activation="tanh", dropout=0.10)(inventory_stream)
    full_branch = SimpleRNN(32, activation="tanh", dropout=0.10)(inputs)

    combined = Concatenate()([demand_branch, inventory_branch, full_branch])
    dense = Dense(32, activation="relu")(combined)
    dense = Dropout(0.20)(dense)
    outputs = Dense(target_hours, activation="sigmoid")(dense)

    model = Model(inputs, outputs, name="dual_stream_simplernn_24h")
    model.compile(
        optimizer=Adam(learning_rate=0.001, clipnorm=1.0),
        loss="binary_crossentropy",
        metrics=["binary_accuracy", AUC(name="auc"), AUC(name="pr_auc", curve="PR")],
    )
    return model

model_6 = build_dual_stream_rnn_24h_model(
    LOOKBACK,
    X_train.shape[2],
    demand_indices,
    inventory_indices,
    TARGET_HOURS,
)
model_6.summary()
```

```python
positive_hours = float(y_train.sum())
total_hours = float(y_train.size)
negative_hours = total_hours - positive_hours
positive_weight = negative_hours / max(positive_hours, 1.0)

# Keras class_weight is awkward for multi-label 24-output targets, so this mirrors
# BCEWithLogitsLoss(pos_weight=...) from the reference file using weighted BCE.
def weighted_binary_crossentropy(pos_weight):
    def loss(y_true, y_pred):
        epsilon = 1e-7
        y_pred = tf.clip_by_value(y_pred, epsilon, 1.0 - epsilon)
        return tf.reduce_mean(
            -(pos_weight * y_true * tf.math.log(y_pred) + (1.0 - y_true) * tf.math.log(1.0 - y_pred))
        )
    return loss

model_6.compile(
    optimizer=Adam(learning_rate=0.001, clipnorm=1.0),
    loss=weighted_binary_crossentropy(positive_weight),
    metrics=["binary_accuracy", AUC(name="auc"), AUC(name="pr_auc", curve="PR")],
)

callbacks_6 = [
    EarlyStopping(
        monitor="val_pr_auc",
        mode="max",
        patience=4,
        restore_best_weights=True,
        verbose=1,
    ),
    ReduceLROnPlateau(
        monitor="val_pr_auc",
        mode="max",
        factor=0.5,
        patience=2,
        min_lr=1e-5,
        verbose=1,
    ),
]

print("Positive hour weight:", positive_weight)
print("Training Variation 6: Reference-Style Dual-Stream SimpleRNN 24H")
history_6 = model_6.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=30,
    batch_size=64,
    callbacks=callbacks_6,
    verbose=1,
)
```
